```python
import time
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
data = pd.read_csv("diabetes - diabetes.csv")

X = data.drop("Outcome", axis=1)
y = data["Outcome"]

print("Dataset Shape:", X.shape)
print("\nFirst 5 Rows:")
display(data.head())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Feature Scaling Completed Successfully!")
```

```python
model = LogisticRegression(max_iter=1000)

start = time.time()
model.fit(X_train_scaled, y_train)
train_original = time.time() - start

start = time.time()
pred_original = model.predict(X_test_scaled)
test_original = time.time() - start

acc_original = accuracy_score(y_test, pred_original)

print("Original Dataset Accuracy:", round(acc_original * 100, 2), "%")
print("Training Time:", round(train_original, 6), "seconds")
print("Prediction Time:", round(test_original, 6), "seconds")
```

```python
pca = PCA(n_components=5)

X_train_pca = pca.fit_transform(X_train_scaled)
X_test_pca = pca.transform(X_test_scaled)

model_pca = LogisticRegression(max_iter=1000)

start = time.time()
model_pca.fit(X_train_pca, y_train)
train_pca = time.time() - start

start = time.time()
pred_pca = model_pca.predict(X_test_pca)
test_pca = time.time() - start

acc_pca = accuracy_score(y_test, pred_pca)

print("PCA Accuracy:", round(acc_pca * 100, 2), "%")
print("Training Time:", round(train_pca, 6), "seconds")
print("Prediction Time:", round(test_pca, 6), "seconds")
print("Reduced Features:", X_train_pca.shape[1])
```

```python
lda = LinearDiscriminantAnalysis(n_components=1)

X_train_lda = lda.fit_transform(X_train_scaled, y_train)
X_test_lda = lda.transform(X_test_scaled)

model_lda = LogisticRegression(max_iter=1000)

start = time.time()
model_lda.fit(X_train_lda, y_train)
train_lda = time.time() - start

start = time.time()
pred_lda = model_lda.predict(X_test_lda)
test_lda = time.time() - start

acc_lda = accuracy_score(y_test, pred_lda)

print("LDA Accuracy:", round(acc_lda * 100, 2), "%")
print("Training Time:", round(train_lda, 6), "seconds")
print("Prediction Time:", round(test_lda, 6), "seconds")
print("Reduced Features:", X_train_lda.shape[1])
```

```python
results = pd.DataFrame({
    "Method": ["Original", "PCA", "LDA"],
    "Features": [8, 5, 1],
    "Accuracy (%)": [
        round(acc_original * 100, 2),
        round(acc_pca * 100, 2),
        round(acc_lda * 100, 2)
    ],
    "Training Time (sec)": [
        round(train_original, 6),
        round(train_pca, 6),
        round(train_lda, 6)
    ],
    "Prediction Time (sec)": [
        round(test_original, 6),
        round(test_pca, 6),
        round(test_lda, 6)
    ]
})

display(results)
```
