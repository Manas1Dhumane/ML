```python
# Import required libraries

import pandas as pd
import numpy as np

# Import preprocessing tools from Scikit-learn
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
```

```python
# Load Titanic dataset

url = "titanic - titanic.csv"

df = pd.read_csv(url)
```

```python
# Display the first five rows of the dataset

df.head()
```

```python
# Display dataset information such as columns, data types and non-null values

df.info()
```

```python
# Check the number of missing values in each column

df.isnull().sum()
```

```python
# Select only the required columns for feature engineering

df = df[['Pclass', 'Sex', 'Age', 'SibSp',
         'Parch', 'Fare', 'Embarked', 'Survived']]

# Display selected columns

df.head()
```

```python
# Separate input features (X) and target variable (y)

X = df.drop('Survived', axis=1)

y = df['Survived']

# X contains input features
# y contains the target variable
```

```python
# Select numerical columns

numeric_features = ['Age', 'Fare', 'SibSp', 'Parch']
```

```python
# Select categorical columns

categorical_features = ['Sex', 'Embarked', 'Pclass']
```

```python
# Create numerical preprocessing pipeline

numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler())
])
```

```python
# Create categorical preprocessing pipeline

categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])
```

```python
# Combine numerical and categorical preprocessing

preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ]
)
```

```python
# Split dataset into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
# Apply preprocessing to training data

X_train_processed = preprocessor.fit_transform(X_train)

# Apply the same preprocessing to testing data

X_test_processed = preprocessor.transform(X_test)
```

```python
# Display original and processed dataset shapes

print("Original Shape:", X_train.shape)

print("Processed Shape:", X_train_processed.shape)
```
